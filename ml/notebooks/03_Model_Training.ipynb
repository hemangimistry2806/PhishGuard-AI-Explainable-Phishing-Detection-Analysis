# Machine Learning Model Training

This notebook focuses on training and evaluating machine learning models using the preprocessed PhiUSIIL Phishing URL Dataset.

The dataset preparation, preprocessing, and feature engineering were completed in the previous notebook. In this phase, ensemble learning algorithms are trained and compared to identify the most suitable model for phishing website detection.

```python
import pandas as pd
import joblib

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)

from xgboost import XGBClassifier
```

## Load Preprocessed Dataset

```python
df = pd.read_csv("cleaned_dataset.csv")
```

```python
# Feature Matrix
X = df.drop(columns=['URL','Domain','TLD','Title','label'])

# Target Variable
y = df['label']

# 80:20 Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Samples :", X_train.shape)
print("Testing Samples  :", X_test.shape)
```

# Random Forest Model

```python
rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf.fit(X_train, y_train)
```

```python
y_pred_rf = rf.predict(X_test)
y_prob_rf = rf.predict_proba(X_test)[:,1]
```

```python
print("Accuracy :", accuracy_score(y_test, y_pred_rf))

print("\nClassification Report\n")
print(classification_report(y_test, y_pred_rf))

print("\nConfusion Matrix\n")
print(confusion_matrix(y_test, y_pred_rf))

print("\nROC-AUC :", roc_auc_score(y_test, y_prob_rf))
```

```python
scores_rf = cross_val_score(
    rf,
    X,
    y,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)

print(scores_rf)
print("Mean Accuracy :", scores_rf.mean())
```

# XGBoost Model

```python
xgb = XGBClassifier(
    random_state=42,
    eval_metric="logloss"
)

xgb.fit(X_train, y_train)
```

```python
y_pred_xgb = xgb.predict(X_test)
y_prob_xgb = xgb.predict_proba(X_test)[:,1]
```

```python
print("Accuracy :", accuracy_score(y_test, y_pred_xgb))

print("\nClassification Report\n")
print(classification_report(y_test, y_pred_xgb))

print("\nConfusion Matrix\n")
print(confusion_matrix(y_test, y_pred_xgb))

print("\nROC-AUC :", roc_auc_score(y_test, y_prob_xgb))
```

```python
scores_xgb = cross_val_score(
    xgb,
    X,
    y,
    cv=5,
    scoring="accuracy",
    n_jobs=-1
)

print(scores_xgb)
print("Mean Accuracy :", scores_xgb.mean())
```

# Model Comparison

```python
comparison = pd.DataFrame({
    "Model": ["Random Forest", "XGBoost"],
    "Accuracy": [
        accuracy_score(y_test, y_pred_rf),
        accuracy_score(y_test, y_pred_xgb)
    ],
    "ROC-AUC": [
        roc_auc_score(y_test, y_prob_rf),
        roc_auc_score(y_test, y_prob_xgb)
    ],
    "5-Fold Cross Validation": [
        scores_rf.mean(),
        scores_xgb.mean()
    ]
})

comparison
```

# Save Final Model

The XGBoost model is selected as the final machine learning model due to its excellent predictive performance and strong generalization capability. The trained model is saved for further Explainable AI (SHAP) analysis and integration into the PhishGuard AI framework.

```python
import os

# Create models folder if it doesn't exist
os.makedirs("../models", exist_ok=True)

# Save trained XGBoost model
joblib.dump(xgb, "../models/xgboost_model.pkl")

print("XGBoost model saved successfully!")
```

# Conclusion

Random Forest and XGBoost were successfully trained and evaluated using the preprocessed PhiUSIIL Phishing URL Dataset.

Although Random Forest achieved perfect performance on the experimental dataset, XGBoost was selected as the final model because it demonstrated nearly identical predictive performance while providing better generalization capability and being widely adopted in phishing detection research.
